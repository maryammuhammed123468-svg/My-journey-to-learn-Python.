```
# for while
# Count Down
count = 15
while count >=1:
  print(count)
  count -= 1
print("Blast Off!")
```

```
# for while
# ATM Password
correct_pin = "1235"
attempts = 3
while attempts > 0 :
  pin = input(f"Enter PIN ({attempts} attempts left):")
  if pin == correct_pin:
    print("Access Granted")
    break
  else:
    print("Incorrect PIN")
    attempts -= 1
  if attempts == 0:
    print("Account Locked")
```

```
# for while
# Leaderboard
players = []
count = 1
while count <= 5:
  name = input(f"Enter name for player {count}:")
  score = int(input(f"Enter Score for {name}:"))
  players.append((score, name))
  count += 1
  players.sort(reverse = True)
print("\n --- Leaderboard ---")
rank = 1
idx = 0
while idx <len(players):
  print(f"{rank}. {players[idx][1]} ({players[idx][0]})")
  rank += 1
  idx += 1
```
